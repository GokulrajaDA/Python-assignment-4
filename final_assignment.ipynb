```python
# 3: Text Cleaning

def clean_text(text):
    # Convert to lowercase
    text = text.lower()
    # Remove...
    text = text.replace('...', ' ')
    # Remove punctuation
    for punct in string.punctuation:
        text = text.replace(punct, ' ')
    # Multiple spaces -> single space
    text = ' '.join(text.split())
    # Remove leading/trailing
    text = text.strip()
    # Slang replacement
    slang_dict = {'ok': 'okay', 'u': 'you', 'ur': 'your', 'gr8': 'great', 'grt': 'great'}
    words = text.split()
    corrected = [slang_dict.get(w, w) for w in words]
    return ' '.join(corrected)

cleaned_descriptions = [clean_text(d) for d in ticket_data['Issue_Description']]
ticket_data['Issue_Description'] = cleaned_descriptions

print("\n========== STEP 3: Cleaned Descriptions ==========")
for d in cleaned_descriptions:
    print(d)
```

```python
# 4: Keyword-Based Insights

def count_tickets_with_word(word):
    word = word.lower()
    count = 0
    for desc in ticket_data['Issue_Description']:
        if word in desc.split(): # exact word match, case-insensitive
            count += 1
    return count

print("\n========== STEP 4: Keyword Insights ==========")
print(f"Tickets containing 'poor': {count_tickets_with_word('poor')}")
print(f"Tickets containing 'good': {count_tickets_with_word('good')}")
print(f"Tickets containing 'slow': {count_tickets_with_word('slow')}")
print(f"Tickets containing 'excellent': {count_tickets_with_word('excellent')}")
```
